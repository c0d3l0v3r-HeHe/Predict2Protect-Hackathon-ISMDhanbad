````python
# Data Preprocessing Pipeline

# This section creates a reusable preprocessing pipeline that:
# 1. Loads data (train or test)
# 2. Handles class imbalance using **Stratified Sampling** (for training data)
# 3. Applies **StandardScaler** to normalize features
# 4. Prepares sequences for padding (to be discussed)
# 5. Saves processed data as compressed numpy arrays
````

````python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import os

print("Preprocessing libraries imported successfully!")
````

## Step 1: Define Preprocessing Function

````python
def preprocess_data(df, is_train=True, scaler=None, apply_stratified_sampling=False):
    """
    Preprocessing pipeline for bankruptcy prediction data
    
    Parameters:
    -----------
    df : pd.DataFrame
        Input dataframe (train or test)
    is_train : bool
        Whether this is training data (affects stratified sampling)
    scaler : StandardScaler or None
        Pre-fitted scaler (for test data). If None, a new scaler is fitted.
    apply_stratified_sampling : bool
        Whether to apply stratified sampling to balance classes (only for train)
    
    Returns:
    --------
    X : dict
        Dictionary with company_name as keys, each containing:
        - 'sequences': np.array of shape (num_years, num_features)
        - 'years': list of fiscal years
    y : dict
        Dictionary with company_name as keys, values are status labels
    scaler : StandardScaler
        Fitted scaler object
    metadata : dict
        Additional information about preprocessing
    """
    
    print("="*60)
    print("PREPROCESSING PIPELINE")
    print("="*60)
    
    # 1. Identify feature columns (financial ratios)
    feature_cols = [col for col in df.columns if col.startswith((
        'Net_', 'Total_', 'Working_', 'Current_', 'Return_', 'Retained_', 
        'EBIT_', 'Book_', 'Sales_', 'Equity_', 'Depreciation_'
    ))]
    
    print(f"\n✓ Identified {len(feature_cols)} financial ratio features")
    
    # 2. Handle Stratified Sampling (only for training data with imbalance)
    if is_train and apply_stratified_sampling and 'status_label' in df.columns:
        print("\n" + "-"*60)
        print("STEP 1: STRATIFIED SAMPLING FOR CLASS BALANCE")
        print("-"*60)
        
        # Get company-level labels
        company_labels = df.groupby('company_name')['status_label'].first()
        
        print(f"\nBefore stratified sampling:")
        print(f"  Total companies: {len(company_labels)}")
        print(f"  Class distribution:")
        print(company_labels.value_counts())
        
        # Get companies for each class
        alive_companies = company_labels[company_labels == 'alive'].index.tolist()
        failed_companies = company_labels[company_labels == 'failed'].index.tolist()
        
        # Undersample majority class to match minority class
        np.random.seed(42)
        sampled_alive = np.random.choice(alive_companies, size=len(failed_companies), replace=False)
        balanced_companies = list(sampled_alive) + failed_companies
        
        # Filter dataframe to only include balanced companies
        df = df[df['company_name'].isin(balanced_companies)].copy()
        
        print(f"\nAfter stratified sampling (undersampling):")
        print(f"  Total companies: {len(balanced_companies)}")
        print(f"  Alive companies: {len(sampled_alive)}")
        print(f"  Failed companies: {len(failed_companies)}")
        print(f"  Rows in dataset: {len(df)}")
        print(f"  ✓ Balanced class distribution achieved!")
    
    # 3. Scale Features
    print("\n" + "-"*60)
    print("STEP 2: FEATURE SCALING (StandardScaler)")
    print("-"*60)
    
    if scaler is None:
        # Fit new scaler on training data
        scaler = StandardScaler()
        df[feature_cols] = scaler.fit_transform(df[feature_cols])
        print(f"✓ Fitted StandardScaler on {len(feature_cols)} features")
    else:
        # Use pre-fitted scaler for test data
        df[feature_cols] = scaler.transform(df[feature_cols])
        print(f"✓ Applied pre-fitted scaler on {len(feature_cols)} features")
    
    # 4. Organize data by company (sequences)
    print("\n" + "-"*60)
    print("STEP 3: ORGANIZE INTO SEQUENCES")
    print("-"*60)
    
    X = {}  # Dictionary: company_name -> sequence data
    y = {}  # Dictionary: company_name -> label
    
    for company_name, group in df.groupby('company_name'):
        # Sort by fiscal year
        group = group.sort_values('Fiscal_Year')
        
        # Extract sequence of features
        sequence = group[feature_cols].values  # Shape: (num_years, num_features)
        years = group['Fiscal_Year'].tolist()
        
        X[company_name] = {
            'sequences': sequence,
            'years': years
        }
        
        # Extract label (same for all years)
        if 'status_label' in df.columns:
            y[company_name] = group['status_label'].iloc[0]
    
    print(f"✓ Created sequences for {len(X)} companies")
    
    # 5. Metadata
    sequence_lengths = [X[comp]['sequences'].shape[0] for comp in X.keys()]
    metadata = {
        'num_companies': len(X),
        'num_features': len(feature_cols),
        'feature_names': feature_cols,
        'min_seq_length': min(sequence_lengths),
        'max_seq_length': max(sequence_lengths),
        'mean_seq_length': np.mean(sequence_lengths),
        'std_seq_length': np.std(sequence_lengths)
    }
    
    print(f"\n{'='*60}")
    print("PREPROCESSING SUMMARY")
    print("="*60)
    print(f"Number of companies: {metadata['num_companies']}")
    print(f"Number of features: {metadata['num_features']}")
    print(f"Sequence length - Min: {metadata['min_seq_length']}, Max: {metadata['max_seq_length']}, Mean: {metadata['mean_seq_length']:.2f}")
    
    if y:
        label_counts = pd.Series(y).value_counts()
        print(f"\nLabel distribution:")
        print(label_counts)
    
    return X, y, scaler, metadata
````

## Step 2: Apply Preprocessing to Training Data

````python
# Apply preprocessing with stratified sampling to balance classes
X_train, y_train, train_scaler, train_metadata = preprocess_data(
    df=train_df.copy(),
    is_train=True,
    scaler=None,
    apply_stratified_sampling=True  # Enable stratified sampling for class balance
)

print("\n" + "="*60)
print("TRAINING DATA PREPROCESSING COMPLETE!")
print("="*60)
````

## Step 4: Save Preprocessed Data as Compressed NumPy Arrays

Before we save, we need to discuss **sequence padding** strategy since companies have variable-length sequences (1-20 years).

````python
def save_preprocessed_data(X, y, filename, metadata=None):
    """
    Save preprocessed sequences as compressed numpy arrays
    
    Parameters:
    -----------
    X : dict
        Dictionary with company sequences
    y : dict
        Dictionary with company labels
    filename : str
        Output filename (without extension)
    metadata : dict
        Metadata to save
    """
    
    # Convert to lists for saving
    company_names = list(X.keys())
    sequences = [X[comp]['sequences'] for comp in company_names]
    years = [X[comp]['years'] for comp in company_names]
    labels = [y[comp] for comp in company_names] if y else None
    
    # Create output directory if it doesn't exist
    os.makedirs('preprocessed_data', exist_ok=True)
    
    # Save as compressed numpy file
    save_dict = {
        'company_names': np.array(company_names, dtype=object),
        'sequences': np.array(sequences, dtype=object),  # Variable length sequences
        'years': np.array(years, dtype=object),
        'metadata': metadata
    }
    
    if labels is not None:
        save_dict['labels'] = np.array(labels, dtype=object)
    
    output_path = f'preprocessed_data/{filename}.npz'
    np.savez_compressed(output_path, **save_dict)
    
    print(f"✓ Saved preprocessed data to: {output_path}")
    print(f"  - Number of companies: {len(company_names)}")
    print(f"  - File size: {os.path.getsize(output_path) / 1024:.2f} KB")
    
    return output_path
````

````python
# Save training data
train_path = save_preprocessed_data(
    X=X_train,
    y=y_train,
    filename='train_preprocessed',
    metadata=train_metadata
)

# Save test data (if available)
if X_test is not None:
    test_path = save_preprocessed_data(
        X=X_test,
        y=y_test,
        filename='test_preprocessed',
        metadata=test_metadata
    )
    
print("\n" + "="*60)
print("ALL DATA SAVED SUCCESSFULLY!")
print("="*60)
````

# 🔍 Discussion: Sequence Padding Strategies

## Current Situation
- **Variable sequence lengths**: Companies have between **1 to 20 years** of data
- **Mean sequence length**: ~8.84 years
- **Standard deviation**: ~5.63 years
- After stratified sampling: **974 companies** (487 alive, 487 failed)

## Padding Options for Sequential Models (LSTM/GRU/Transformer)

### **Option 1: Pad to Maximum Length (20 years)** ⭐ Recommended
**Pros:**
- No information loss - preserves all temporal data
- Standard approach for LSTM/GRU
- Works well with masking layers in Keras/PyTorch
- Best for capturing long-term temporal patterns

**Cons:**
- Memory intensive (especially for companies with 1-2 years)
- Lots of padding for short sequences (50+ companies have ≤3 years)
- Padding might introduce noise

**Implementation:**
```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
# Pad sequences to max_length=20, pad with zeros at the beginning
X_padded = pad_sequences(sequences, maxlen=20, padding='pre', dtype='float32')
```

### **Option 2: Pad to Percentile (e.g., 90th percentile)** 
**Pros:**
- More memory efficient
- Reduces excessive padding
- Captures most sequences fully

**Cons:**
- Need to truncate sequences longer than threshold
- Might lose important temporal information for long sequences

**Implementation:**
```python
# Use 90th percentile (~16 years) or 75th percentile (~13 years)
threshold = int(np.percentile(sequence_lengths, 90))
```

### **Option 3: Bucketing - Group by Sequence Length**
**Pros:**
- Very memory efficient
- Minimal padding per bucket
- Faster training within each bucket

**Cons:**
- Complex implementation
- Need multiple models or dynamic batching
- Harder to manage

**Example Buckets:**
- Bucket 1: 1-5 years (pad to 5)
- Bucket 2: 6-12 years (pad to 12)
- Bucket 3: 13-20 years (pad to 20)

### **Option 4: Use Masking Layers** ⭐ Best Practice
**Combine with Option 1 or 2:**
```python
from tensorflow.keras.layers import Masking

# Add masking layer - tells model to ignore padded values
model.add(Masking(mask_value=0.0))  # Assuming padding with zeros
model.add(LSTM(128, return_sequences=True))
```

This way the model learns to ignore padded positions!

**For LSTM/GRU/Transformer Models:**
1. **Pad to max_length=20** (Option 1)
2. **Use Masking layer** (Option 4) to handle variable lengths
3. Pad with **zeros** at the **beginning** (`padding='pre'`)
4. This preserves all temporal information while telling the model which positions are real vs. padded

- Extract aggregated features (mean, std, trend, etc.)

## Step 1: Prepare Padded Sequences for LSTM

````python
def prepare_padded_data(X_dict, y_dict, max_length=20):
    """
    Convert variable-length sequences to padded arrays for LSTM
    
    Parameters:
    -----------
    X_dict : dict
        Dictionary with company sequences
    y_dict : dict
        Dictionary with company labels
    max_length : int
        Maximum sequence length for padding
    
    Returns:
    --------
    X_padded : np.array
        Padded sequences (num_samples, max_length, num_features)
    y_array : np.array
        Labels as array
    company_names : list
        List of company names (for reference)
    """
    
    company_names = list(X_dict.keys())
    sequences = []
    labels = []
    
    for company in company_names:
        sequences.append(X_dict[company]['sequences'])
        labels.append(y_dict[company])
    
    # Pad sequences - pad with zeros at the end (post-padding)
    # Use 'post' padding for cuDNN compatibility with masking
    X_padded = pad_sequences(
        sequences, 
        maxlen=max_length, 
        padding='post',  # Pad at the end (required for cuDNN + masking)
        dtype='float32',
        value=0.0  # Pad with zeros
    )
    
    # Convert labels to binary (0 = alive, 1 = failed)
    label_map = {'alive': 0, 'failed': 1}
    y_array = np.array([label_map[label] for label in labels])
    
    print("="*60)
    print("PADDED DATA PREPARATION")
    print("="*60)
    print(f"Number of samples: {len(X_padded)}")
    print(f"Padded shape: {X_padded.shape}")
    print(f"  - Max sequence length: {X_padded.shape[1]}")
    print(f"  - Number of features: {X_padded.shape[2]}")
    print(f"\nLabel distribution:")
    print(f"  - Alive (0): {np.sum(y_array == 0)}")
    print(f"  - Failed (1): {np.sum(y_array == 1)}")
    print(f"  - Balance: {np.sum(y_array == 0) / np.sum(y_array == 1):.2f}:1")
    
    return X_padded, y_array, company_names

# Prepare training data
X_train_padded, y_train_array, train_company_names = prepare_padded_data(
    X_train, y_train, max_length=20
)

print("\n✓ Training data padded and ready for LSTM!")
````

## Step 3: Build LSTM Model with Masking

````python
def build_lstm_model(input_shape, lstm_units=64, dropout_rate=0.3):
    """
    Build LSTM model with Masking layer for variable-length sequences
    
    Parameters:
    -----------
    input_shape : tuple
        (sequence_length, num_features)
    lstm_units : int
        Number of LSTM units
    dropout_rate : float
        Dropout rate for regularization
    
    Returns:
    --------
    model : keras.Model
        Compiled LSTM model
    """
    
    model = Sequential([
        # Masking layer - tells model to ignore padded values (zeros)
        Masking(mask_value=0.0, input_shape=input_shape),
        
        # Bidirectional LSTM - processes sequences in both directions
        Bidirectional(LSTM(lstm_units, return_sequences=True)),
        Dropout(dropout_rate),
        
        # Second LSTM layer
        Bidirectional(LSTM(lstm_units // 2)),
        Dropout(dropout_rate),
        
        # Dense layers
        Dense(32, activation='relu'),
        Dropout(dropout_rate),
        
        Dense(16, activation='relu'),
        
        # Output layer - binary classification
        Dense(1, activation='sigmoid')
    ])
    
    # Compile model
    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=[
            'accuracy',
            tf.keras.metrics.AUC(name='auc'),
            tf.keras.metrics.Precision(name='precision'),
            tf.keras.metrics.Recall(name='recall')
        ]
    )
    
    return model

# Build model
input_shape = (X_train_split.shape[1], X_train_split.shape[2])  # (20, 18)
lstm_model = build_lstm_model(input_shape, lstm_units=64, dropout_rate=0.3)

print("="*60)
print("LSTM MODEL ARCHITECTURE")
print("="*60)
lstm_model.summary()
print("\n✓ Model built successfully!")
````

````python
from tensorflow.keras.layers import GRU

# Store results for comparison
model_results = []

def train_and_evaluate_model(model, model_name, X_train, y_train, X_val, y_val, epochs=100):
    """
    Train a model and return evaluation metrics
    """
    print(f"\n{'='*60}")
    print(f"Training: {model_name}")
    print("="*60)
    
    # Callbacks
    early_stop = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True, verbose=0)
    reduce_lr_callback = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=0)
    
    # Train
    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=32,
        callbacks=[early_stop, reduce_lr_callback],
        verbose=0
    )
    
    # Evaluate
    y_pred_proba = model.predict(X_val, verbose=0)
    y_pred = (y_pred_proba > 0.5).astype(int).flatten()
    
    # Metrics
    accuracy = np.mean(y_pred == y_val)
    auc = roc_auc_score(y_val, y_pred_proba)
    f1 = f1_score(y_val, y_pred)
    mcc_score = matthews_corrcoef(y_val, y_pred)
    cm = confusion_matrix(y_val, y_pred)
    tn, fp, fn, tp = cm.ravel()
    sensitivity = tp / (tp + fn)
    specificity = tn / (tn + fp)
    
    # Store results
    results = {
        'model_name': model_name,
        'accuracy': accuracy,
        'auc': auc,
        'f1': f1,
        'mcc': mcc_score,
        'sensitivity': sensitivity,
        'specificity': specificity,
        'epochs': len(hist.history['loss']),
        'model': model,
        'history': hist
    }
    
    print(f"✓ {model_name}")
    print(f"  Accuracy: {accuracy:.4f} | AUC: {auc:.4f} | F1: {f1:.4f}")
    print(f"  Sensitivity: {sensitivity:.4f} | Specificity: {specificity:.4f}")
    print(f"  Epochs: {len(hist.history['loss'])}")
    
    return results

print("✓ Training helper function defined!")
````

## Step 3: Optimize Threshold for Macro-F1 and Recall

Since the evaluation metric is Macro-F1 and recall on bankruptcy class is secondary, we should optimize the threshold to maximize these metrics on the validation set.

````python
# Load validation data from training set
loaded_train = np.load('preprocessed_data/train_preprocessed.npz', allow_pickle=True)
train_company_names = loaded_train['company_names'].tolist()
train_sequences = loaded_train['sequences']
train_labels = loaded_train['labels']

# Pad training sequences
train_seq_list = [seq for seq in train_sequences]
X_train_padded = pad_sequences(
    train_seq_list,
    maxlen=20,
    padding='post',
    dtype='float32',
    value=0.0
)

# Convert labels to binary
label_map = {'alive': 0, 'failed': 1}
y_train_array = np.array([label_map[label] for label in train_labels])

# Split into train/val (same as before: 80/20)
from sklearn.model_selection import train_test_split
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_padded, 
    y_train_array,
    test_size=0.2,
    random_state=42,
    stratify=y_train_array
)

print(f"✓ Loaded validation set: {X_val_split.shape[0]} samples")

# Use validation set to find optimal threshold
# Get predictions on validation set from best model
val_pred_proba_best = best_model_loaded.predict(X_val_split, verbose=0).flatten()

# Test different thresholds
thresholds_to_test = np.arange(0.1, 0.9, 0.02)
threshold_results = []

for thresh in thresholds_to_test:
    val_pred = (val_pred_proba_best >= thresh).astype(int)
    
    # Calculate macro F1
    macro_f1 = f1_score(y_val_split, val_pred, average='macro')
    
    # Calculate recall for bankruptcy class (label=1, failed)
    cm = confusion_matrix(y_val_split, val_pred)
    if cm.shape == (2, 2):
        tn, fp, fn, tp = cm.ravel()
        recall_bankruptcy = tp / (tp + fn) if (tp + fn) > 0 else 0
    else:
        recall_bankruptcy = 0
    
    # Calculate per-class F1
    f1_per_class = f1_score(y_val_split, val_pred, average=None)
    
    threshold_results.append({
        'threshold': thresh,
        'macro_f1': macro_f1,
        'recall_bankruptcy': recall_bankruptcy,
        'f1_alive': f1_per_class[0] if len(f1_per_class) > 0 else 0,
        'f1_failed': f1_per_class[1] if len(f1_per_class) > 1 else 0
    })

# Convert to DataFrame
threshold_df = pd.DataFrame(threshold_results)

# Find best threshold (prioritize macro-F1, then recall)
best_idx = threshold_df['macro_f1'].idxmax()
optimal_threshold = threshold_df.loc[best_idx, 'threshold']
best_macro_f1 = threshold_df.loc[best_idx, 'macro_f1']
best_recall = threshold_df.loc[best_idx, 'recall_bankruptcy']

print("="*60)
print("THRESHOLD OPTIMIZATION RESULTS")
print("="*60)
print(f"\nOptimal Threshold: {optimal_threshold:.3f}")
print(f"Expected Macro-F1: {best_macro_f1:.4f}")
print(f"Expected Recall (Bankruptcy): {best_recall:.4f}")

# Show top 5 thresholds
print("\nTop 5 Thresholds by Macro-F1:")
print(threshold_df.nlargest(5, 'macro_f1')[['threshold', 'macro_f1', 'recall_bankruptcy', 'f1_alive', 'f1_failed']].to_string(index=False))

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(threshold_df['threshold'], threshold_df['macro_f1'], label='Macro-F1', linewidth=2)
axes[0].axvline(optimal_threshold, color='red', linestyle='--', label=f'Optimal ({optimal_threshold:.3f})')
axes[0].set_xlabel('Threshold')
axes[0].set_ylabel('Macro-F1 Score')
axes[0].set_title('Threshold vs Macro-F1 Score')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(threshold_df['threshold'], threshold_df['recall_bankruptcy'], label='Recall (Bankruptcy)', color='orange', linewidth=2)
axes[1].plot(threshold_df['threshold'], threshold_df['f1_alive'], label='F1 (Alive)', color='green', linewidth=2, alpha=0.5)
axes[1].plot(threshold_df['threshold'], threshold_df['f1_failed'], label='F1 (Failed)', color='red', linewidth=2, alpha=0.5)
axes[1].axvline(optimal_threshold, color='red', linestyle='--', label=f'Optimal ({optimal_threshold:.3f})')
axes[1].set_xlabel('Threshold')
axes[1].set_ylabel('Score')
axes[1].set_title('Threshold vs Individual Metrics')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()

print("\n✓ Threshold optimization complete!")
````

# Option 2: Feature Engineering + Traditional ML

Now let's try a different approach: extracting temporal features from the time series and using traditional ML models (XGBoost, LightGBM, Random Forest). This approach:
1. **Aggregates time series** into statistical features
2. **Captures trends and volatility** over time
3. **Uses ensemble methods** that handle tabular data well
4. Can potentially **ensemble with LSTM** for best results

## Step 1: Feature Engineering - Extract Temporal Features

````python
def extract_temporal_features(X_dict, feature_names=None):
    """
    Extract comprehensive temporal features from time series data
    
    Features extracted:
    1. Statistical: mean, std, min, max, median, skewness, kurtosis
    2. Trend: linear regression slope, R²
    3. Temporal: first year, last year, year-over-year changes
    4. Volatility: coefficient of variation, range
    5. Moving averages: 3-year, 5-year windows
    6. Ratios: last/first, recent/historic
    
    Parameters:
    -----------
    X_dict : dict
        Dictionary with company sequences
    feature_names : list
        Names of features (for naming extracted features)
    
    Returns:
    --------
    features_df : pd.DataFrame
        DataFrame with extracted features for each company
    """
    from scipy.stats import skew, kurtosis, linregress
    
    if feature_names is None:
        feature_names = [f'X{i}' for i in range(1, 19)]
    
    print("="*60)
    print("EXTRACTING TEMPORAL FEATURES")
    print("="*60)
    
    features_list = []
    company_names = list(X_dict.keys())
    
    for company in company_names:
        sequences = X_dict[company]['sequences']
        num_years = X_dict[company]['num_years']
        
        company_features = {}
        company_features['company_name'] = company
        company_features['num_years'] = num_years
        
        # For each financial ratio
        for i, feat_name in enumerate(feature_names):
            time_series = sequences[:, i]
            
            # 1. Basic Statistics
            company_features[f'{feat_name}_mean'] = np.mean(time_series)
            company_features[f'{feat_name}_std'] = np.std(time_series)
            company_features[f'{feat_name}_min'] = np.min(time_series)
            company_features[f'{feat_name}_max'] = np.max(time_series)
            company_features[f'{feat_name}_median'] = np.median(time_series)
            company_features[f'{feat_name}_range'] = np.max(time_series) - np.min(time_series)
            
            # 2. Distribution Shape
            company_features[f'{feat_name}_skew'] = skew(time_series)
            company_features[f'{feat_name}_kurtosis'] = kurtosis(time_series)
            
            # 3. Volatility Measures
            if company_features[f'{feat_name}_mean'] != 0:
                company_features[f'{feat_name}_cv'] = company_features[f'{feat_name}_std'] / abs(company_features[f'{feat_name}_mean'])
            else:
                company_features[f'{feat_name}_cv'] = 0
            
            # 4. Trend Analysis (Linear Regression Slope)
            if num_years > 1:
                x = np.arange(num_years)
                slope, intercept, r_value, p_value, std_err = linregress(x, time_series)
                company_features[f'{feat_name}_trend_slope'] = slope
                company_features[f'{feat_name}_trend_r2'] = r_value**2
            else:
                company_features[f'{feat_name}_trend_slope'] = 0
                company_features[f'{feat_name}_trend_r2'] = 0
            
            # 5. First and Last Year Values
            company_features[f'{feat_name}_first'] = time_series[0]
            company_features[f'{feat_name}_last'] = time_series[-1]
            
            # 6. Year-over-Year Changes
            if num_years > 1:
                yoy_changes = np.diff(time_series)
                company_features[f'{feat_name}_yoy_mean'] = np.mean(yoy_changes)
                company_features[f'{feat_name}_yoy_std'] = np.std(yoy_changes)
                company_features[f'{feat_name}_last_yoy_change'] = time_series[-1] - time_series[-2] if num_years > 1 else 0
            else:
                company_features[f'{feat_name}_yoy_mean'] = 0
                company_features[f'{feat_name}_yoy_std'] = 0
                company_features[f'{feat_name}_last_yoy_change'] = 0
            
            # 7. Last/First Ratio (overall change)
            if time_series[0] != 0:
                company_features[f'{feat_name}_last_first_ratio'] = time_series[-1] / time_series[0]
            else:
                company_features[f'{feat_name}_last_first_ratio'] = 0
            
            # 8. Moving Averages (if enough data)
            if num_years >= 3:
                company_features[f'{feat_name}_ma3'] = np.mean(time_series[-3:])
            else:
                company_features[f'{feat_name}_ma3'] = np.mean(time_series)
            
            if num_years >= 5:
                company_features[f'{feat_name}_ma5'] = np.mean(time_series[-5:])
            else:
                company_features[f'{feat_name}_ma5'] = np.mean(time_series)
            
            # 9. Recent vs Historic (last 3 years vs first 3 years)
            if num_years >= 6:
                recent_mean = np.mean(time_series[-3:])
                historic_mean = np.mean(time_series[:3])
                if historic_mean != 0:
                    company_features[f'{feat_name}_recent_historic_ratio'] = recent_mean / historic_mean
                else:
                    company_features[f'{feat_name}_recent_historic_ratio'] = 0
            else:
                company_features[f'{feat_name}_recent_historic_ratio'] = 1
        
        features_list.append(company_features)
    
    features_df = pd.DataFrame(features_list)
    
    print(f"\n✓ Extracted features from {len(company_names)} companies")
    print(f"✓ Total features: {len(features_df.columns) - 1} (excluding company_name)")
    print(f"✓ Features per ratio: {(len(features_df.columns) - 2) // len(feature_names)}")  # -2 for company_name and num_years
    
    return features_df

print("✓ Feature extraction function defined!")
````

### Model 1: XGBoost

````python
# Train XGBoost model
print("="*60)
print("TRAINING XGBOOST MODEL")
print("="*60)

xgb_model = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    gamma=0.1,
    min_child_weight=1,
    reg_alpha=0.1,
    reg_lambda=1.0,
    random_state=42,
    eval_metric='logloss',
    tree_method='hist'
)

xgb_model.fit(
    X_train_ml_split, 
    y_train_ml_split,
    eval_set=[(X_val_ml_split, y_val_ml_split)],
    verbose=50
)

# Predictions
y_val_pred_xgb_proba = xgb_model.predict_proba(X_val_ml_split)[:, 1]
y_val_pred_xgb = (y_val_pred_xgb_proba >= 0.5).astype(int)

# Metrics
xgb_accuracy = accuracy_score(y_val_ml_split, y_val_pred_xgb)
xgb_f1 = f1_score(y_val_ml_split, y_val_pred_xgb)
xgb_auc = roc_auc_score(y_val_ml_split, y_val_pred_xgb_proba)
xgb_macro_f1 = f1_score(y_val_ml_split, y_val_pred_xgb, average='macro')
xgb_cm = confusion_matrix(y_val_ml_split, y_val_pred_xgb)

print("\n" + "="*60)
print("XGBOOST PERFORMANCE")
print("="*60)
print(f"Accuracy:  {xgb_accuracy:.4f}")
print(f"F1-Score:  {xgb_f1:.4f}")
print(f"Macro-F1:  {xgb_macro_f1:.4f}")
print(f"AUC-ROC:   {xgb_auc:.4f}")
print(f"\nConfusion Matrix:")
print(xgb_cm)

if xgb_cm.shape == (2, 2):
    tn, fp, fn, tp = xgb_cm.ravel()
    xgb_recall = tp / (tp + fn)
    xgb_specificity = tn / (tn + fp)
    print(f"\nRecall (Bankruptcy): {xgb_recall:.4f}")
    print(f"Specificity:         {xgb_specificity:.4f}")

print("\n✓ XGBoost training complete!")
````
